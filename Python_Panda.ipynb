```python
# Day 4: Working with Conditional Statements (If/Else) for sensor thresholds

# 1. Defining a sensor's current temperature and the safety threshold
current_temperature = 526.4
safety_threshold = 525.0

# 2. Checking the condition
if current_temperature > safety_threshold:
    print("Warning: Sensor temperature has exceeded the safety threshold!")
    print("Current Temp:", current_temperature)
else:
    print("Status: Normal. Sensor is operating within safe limits.")

# 3. Using elif (else if) for multiple conditions
vibration_level = 0.8

if vibration_level < 0.5:
    print("Vibration Status: Low")
elif vibration_level <= 1.0:
    print("Vibration Status: Moderate (Monitor closely)")
else:
    print("Vibration Status: Critical! Immediate inspection required.")
```

```python
# Day 5: Using For Loops to iterate through sensor data

# 1. Defining a list of temperature readings
vibration_levels=[0.1,0.12,0.35,0.81,0.711]
temperature_readings = [520.5, 521.2, 519.8, 526.1, 520.0]
safety_threshold = 525.0

# 2. Iterating through the list using a for loop
for temp in temperature_readings:
    # 3. Checking condition for each individual reading
    if temp > safety_threshold:
        print("Warning: High temperature detected -", temp)
    else:
        print("Normal temperature -", temp)
for vib in vibration_levels:
    if vib > 0.7 :
      print("Critical vibration detected: ",vib)
    else:
      print("Vibration normal: ",vib)
```

```python
# Day 6: Defining Functions (def) for reusable calculations

# 1. Defining a function to calculate total operating cycles
def calculate_total_cycles(previous_cycles, new_cycles):
    total = previous_cycles + new_cycles
    return total

# 2. Calling the function with specific numbers for different engines
engine_1_total = calculate_total_cycles(150, 45)
engine_2_total = calculate_total_cycles(210, 30)

# 3. Printing the results
print("Engine 1 Total Cycles:", engine_1_total)
print("Engine 2 Total Cycles:", engine_2_total)
```

```python
# Day 7: Integration of Functions, Loops, and Conditionals

# 1. Defining a function to filter critical temperatures
def filter_critical_temperatures(temperatures_list, threshold):
    critical_values = [] # Creating an empty list to store dangerous values

    # 2. Looping through the provided list
    for temp in temperatures_list:
        # 3. Checking the condition
        if temp > threshold:
            critical_values.append(temp) # Adding to the list if the condition is met

    return critical_values # Returning the final filtered list

# 4. Creating a list of sensor readings and defining the limit
sensor_readings = [520.1, 526.4, 522.3, 528.0, 519.9, 530.1]
safety_limit = 525.0

# 5. Calling the function and printing the result
dangerous_temps = filter_critical_temperatures(sensor_readings, safety_limit)
print("All Sensor Readings:", sensor_readings)
print("Critical Temperatures Detected:", dangerous_temps)
```

```python
import pandas as pd
import numpy as np

# Day 11: Handling Missing Data

# 1. Dataset with missing values (np.nan simulates missing sensor records)
sensor_data = {
    "engine_id": [1, 1, 2, 2, 3],
    "sensor_temp": [518.67, np.nan, 518.62, 519.85, np.nan]
}
df = pd.DataFrame(sensor_data)
print("Original Data:\n", df, "\n")

# 2. Strategy A: Drop any row containing missing data
df_dropped = df.dropna()
print("Data after dropping NaNs:\n", df_dropped, "\n")

# 3. Strategy B: Fill missing data with a nominal fixed value
df_filled = df.fillna(518.60)
print("Data after filling NaNs:\n", df_filled)

# 4. filling the missing data with the mean of sensor_temp

nominal_temp = df["sensor_temp"].mean()
df_filled_mean = df.fillna(nominal_temp)
print("Data after filling NaNs with mean:\n", df_filled_mean)
```

***Day 12: Grouping Data and Aggregation***

```python
import pandas as pd

# Day 12: Grouping Data (groupby)

# 1. Sample dataset representing multiple engines and their operational cycles
data = {
    "engine_id": [1, 1, 1, 2, 2, 3, 3, 3, 3],
    "cycle": [1, 2, 3, 1, 2, 1, 2, 3, 4],
    "sensor_temp": [518.67, 518.68, 518.69, 519.85, 519.82, 520.10, 520.15, 520.12, 520.18]
}
df = pd.DataFrame(data)
print("Original DataFrame:\n", df, "\n")

# 2. Grouping by engine_id to find the maximum cycle for each engine (End of Life)
# This logic is the core of our RUL calculation phase.
max_cycles = df.groupby("engine_id")["cycle"].max()
print("Maximum cycles per engine (Point of failure):\n", max_cycles)
#3. Task
mean_temp_per_engine = df.groupby("engine_id")["sensor_temp"].mean()
print("\n",mean_temp_per_engine)
```

**Day 13 : Merging Data**

```python
import pandas as pd

# Day 13: Merging DataFrames (pd.merge)

# 1. Main dataset (Sensor readings per cycle)
sensor_data = pd.DataFrame({
    "engine_id": [1, 1, 1, 2, 2],
    "cycle": [1, 2, 3, 1, 2],
    "sensor_temp": [518.67, 518.68, 518.69, 519.85, 519.82]
})
print("Main Data:\n", sensor_data, "\n")

# 2. Max cycles calculated in Day 12 (Point of failure)
max_cycles = pd.DataFrame({
    "engine_id": [1, 2],
    "max_cycle": [3, 2]
})
print("Max Cycles Data:\n", max_cycles, "\n")

# 3. Merging the two tables based on 'engine_id'
merged_df = pd.merge(sensor_data, max_cycles, on="engine_id", how="left")

merged_df["RUL"] = merged_df["max_cycle"] - merged_df["cycle"]
print("Merged DataFrame:\n", merged_df)
```

**Day 14: Rolling Window Operation**

```python
import pandas as pd

# Day 14: Rolling Window Operations (Moving Average)

# 1. Dataset representing sensor readings of a single engine over time
sensor_data = pd.DataFrame({
    "cycle": [1, 2, 3, 4, 5, 6, 7],
    "sensor_temp": [518.6, 520.1, 517.9, 521.5, 519.0, 522.3, 520.8]
})
print("Original Data:\n", sensor_data, "\n")

# 2. Calculating Rolling Mean (Window size = 3)
sensor_data["temp_moving_avg"] = sensor_data["sensor_temp"].rolling(window=3).mean()

# 3. Handling NaN values using backward fill (bfill)
sensor_data["temp_moving_avg"] = sensor_data["temp_moving_avg"].bfill()

print("Data with Rolling Mean and bfill:\n", sensor_data)
```

**Day 15: Data Normalization**

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

# Day 15: Data Normalization

# 1. Dataset with sensors in completely different scales
data = {
    "cycle": [1, 2, 3, 4, 5],
    "sensor_temp": [518.6, 520.1, 517.9, 521.5, 522.3],
    "sensor_pressure": [14.6, 14.1, 14.8, 13.9, 14.5]
}
df = pd.DataFrame(data)
print("Original Data:\n", df, "\n")

# 2. Initializing the Scaler
scaler = MinMaxScaler()

# 3. Scaling a single column (sensor_temp)
df[["sensor_temp_scaled","sensor_pressure_scaled"]] = scaler.fit_transform(df[["sensor_temp","sensor_pressure"]])
print("Data with Scaled Temperature:\n", df)
```
